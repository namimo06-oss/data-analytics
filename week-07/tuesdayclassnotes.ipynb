## Continuing with NumPy 

### Briefly, review hstack/ vstack / dstack

```python
import numpy as np
```

```python
a = [0, 1, 2, 3, 11, 12]
b = [4, 5, 6, 7, 13, 14]

print(np.hstack((a,b))) # equivalent to concatenating on the 0 axis 
print(np.vstack((a,b))) # concatenating on the 1 axis
print(np.dstack((a,b)))  # concatenating on the 2 axis ,it always return in 3 dimension     
```

```python
print(np.concatenate((a,b)))

print(np.stack((a,b)))

print(np.stack((a,b), axis = 1)) # not equivalent to dstack , as dstack will return 3D array

# stacking is adding another dimension 
# concatenate is on the same dimension 
```

```python
# so splitting arrays i breaking arrays apart 
print(np.array_split(a, 2))
```

```python
print(np.array_split(a, 3))
```

```python
print(np.array_split(a, 4)) # allows for an uneven split
```

```python
ab = np.stack((a,b))

print(ab)

print(np.hsplit(ab, 2)) # hsplir works on the axis, is not forgiving of uneven splits
```

```python
ab = np.stack((a,b))

print(ab)

print(np.vsplit(ab, 2))
```

```python
print(np.hsplit(ab, 2))
```

```python
# let's put this into context with some sales data
full_year = np.array([[100, 120, 130, 140, 150, 160, 170, 180, 190, 200, 210, 220],
                     [200, 210, 190, 205, 220, 230, 240, 235, 225, 215, 205, 195],
                     [90, 95, 100, 110, 115, 120, 125, 130, 120, 115, 110, 105]])
# each array is a column
# the "rows" match on th evertical alignment 
```

```python
quarters = np.array_split(full_year, 4, axis = 1)
print(quarters)
q1, q2, q3, q4 = quarters
print(q1)
```

```python
by_product = np.array_split(full_year, 3, axis = 0)

prod_a, prod_b, prod_c = by_product

print(prod_a)
```

## Using where() to apply a condition to an array

```python
### where() let you look for a condition 
```

```python
print(np.where(full_year > 200)) # with only a condition, return an array of matches 
```

```python
print(np.where(full_year > 200, full_year, 'low')) # lets you set a return and not-found value 
```

```python
print(full_year[full_year > 200]) # drop the not matching values 
# filtering with bracket notation return only matches 
```

```python
print(full_year[full_year % 3 == 0]) # % is the reminding 
# will return values divisible by 3 
```

```python
div_by_3 = (full_year % 3 == 0)

print(full_year[div_by_3]) # this is the pattern on pg 15 in the student guide 
```

## Random in NumPy

```python
# when calling the random module in NumPy:
# np.random. [function name]
```

```python
some_ints = np.random.randint(100, size = (5)) # assumes a zero start == from 0 to 100
print(some_ints)
```

```python
more_ints = np.random.randint(50, 100, size = (5,3)) # assumes a zero  == from 0 to 100
print(more_ints)
```

```python
print(np.random.rand(5, 3))
```

```python
# working with choice()
print(some_ints) # starting values to choose from 
print(np.random.choice(some_ints))
```

```python
sales_sample = np.random.choice(str_x, size = (4,10))
print(sales_sample)
```

```python
# choice() can also be used to assign probability values

weigthed_sales_sample = np.random.choice(str_x, size = (4, 10), p = [.1, .2, .2, .2, .1, .05, .05, .05, .05])
print(weighted_sales_sample)
```

## Ufuncs in NumPy

```python
sale_price = np.array([2.99, 12.99, 3.50, 4.50, 17.90, 22.00])

discount_amount = np.array([0, 0, 3.50, .50, 3, 0])

print(sale_price + discount_amount)
```

```python
print(np.add(sale_price, discount_amount))
```

```python
print(np.multiply(sale_price, .8))
```

```python
print(np.round(np.multiply(sale_price, .8), 2))
```

```python
# round() applies "banker's roundind"
# rounding at 5 rounds towars the nearest even value 

apple = np.array((.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6.5))

print(np.round(apple))
```
