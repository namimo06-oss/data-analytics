## Generating Random Choices from an Array

```python
# Create a variable named students and assign it a 1-D array containing the names
# of 12 of your classmates. 8 WILL BE THE NUMBER OF STUDENTS IN MY CLASS so 4 are makeup
students = np.array(["Dahlia", "Brandon", "Edmund", "Destiney", "Luis", "Angel", "Kevin", "Jeremiah", "Hitrik", "Miranda", "Stefanie", "Bess"])
print(students)
```

```python
# Create a variable named call_on and use it to choose one random selection from
# students. Print call_on.
call_on = np.random.choice(students)
print(call_on)
```

```python
# Create a variable named breakout_groups and use it to choose a random
# sampling from students that is 3 rows, 4 columns in size. Print
breakout_groups = np.random.choice(students, size = (3,4))  # ADDDING replce= False will eliminate duplicates
print(breakout_groups)

# This is NOT an effective way to assign breakout groups because
# the same student can appear multiple times and some students
# may not appear at all.
```

```python
# Weighted random selections (5 picks)
p_values = [
    0.30,  # Dahlia (30% chance)
    0.10,  # Brandon
    0.10,  # Edmund
    0.05,  # Destiney
    0.05,  # luis
    0.10,  # Angel
    0.10,  # Kevin
    0.05,  # Jeremiah
    0.05,  # Hritrik
    0.05,  # Miranda
    0.05,  # Stefanie
    0.00   # Bess (0% chance)
]

call_on_weighted = np.random.choice(
    students,
    size=5,
    p=p_values
)

print(call_on_weighted)

# Dahlia has a 30% chance of being selected.
# Bess has a 0% chance and will never be selected.
# Other students have smaller probabilities.
```

## Generating a Normal Distribution 

```python
# Create a variable named bell_curve and assign it a random distribution for
# which the mean (peak of the curve) is 10, the standard deviation is 5, and the size of
# the returned array is 4 rows by 5 columns. Print bell_curve

bell_curve = np.random.normal(
    loc=10,      # mean (center of the curve)
    scale=5,     # standard deviation
    size=(4, 5)  # 4 rows, 5 columns
)

print(bell_curve)
```
