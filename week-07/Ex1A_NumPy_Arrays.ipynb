## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13] , [14, 15, 16]]) # creating a variable as two-dimensional array 
print(arrA) # print result 
```

```python
# loop through the first dimension of arrA /// multiplied by 2
for x in arrA:
    print(x * 2)
```

```python
# loop through the second dimension of arrA /// multiplied by 2 
for x in arrA:
    for y in x:
        print(y * 2)
```

### Difference Between the First Loop and the Second Loop
In the first loop, the loop goes through the outer array only. Each iteration returns one whole row (or sub-list) at a time. 
This means the output shows lists, not individual values.
In the second loop, there is a nested loop. The first loop still goes through each row, but the second loop goes inside each row and 
accesses each individual element. This means the output shows the values one by one instead of lists.

```python
#second dimension with variable loop_result
for x in arrA:
    for y in x:
        loop_result = x*2
print(loop_result)

# Explanation:
# This result is different because loop_result is updated many times inside the loops.
# Each new value replaces the previous one.
# Because print is outside the loops, only the last calculated value is shown.
```

## Joining and Stacking Arrays

```python
# create variable
# axis = 0 so it will add more rows  NOT COLUMNS
joined_array_1 = np.concatenate((arrA, arrA), axis = 0)
print(joined_array_1)
```

```python
# create variable with axis = 1 so columns will be add
joined_array_2 = np.concatenate((arrA, arrA), axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(joined_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

# ValueError: all input arrays must have the same shape 
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
# concatenate joins arrays along axis=0 by default.
# This works because both arrays have the same number of columns.
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)

# ERROR explanation:
# axis=1 means adding columns.
# joined_array_1 has more rows than arrA.
# NumPy cannot match rows, so the operation fails.
# NumPy cannot join these arrays sideways because they do not have the same number of rows.
```

```python
k = np.hstack((arrA, arrA))
n = np.vstack((arrA, arrA))
m = np.dstack((arrA, arrA))

print(k)
print(n)
print(m)

# hstack(): stacks arrays horizontally (adds columns)
# vstack(): stacks arrays vertically (adds rows)
# dstack(): stacks arrays along a third dimension (depth)
```

## Splitting Arrays

```python
splitsky = np.split(joined_array_1, 2)
print(splitsky)

# Result:
# - This produces 2 arrays.
# - Each array is 2-dimensional.
# - Each array has the same shape (half the rows of joined_array_1).
```

```python
spliterino = np.split(joined_array_1, 3)
print(spliterino)

# Result
# Error: array split does not result in an equal division
# This raises an error because joined_array_1 cannot be evenly divided into 3 equal parts.
# The number of rows is not divisible by 3.
```

## Sorting and Filtering Arrays

```python
# create a variable arrB as a 2D array
arrB = np.array([[7,6],
                [13,5],
                [7, 3]])
arrB_sorted = np.sort(arrB) # create another variable and assign it the sorted value of arrB
print(arrB_sorted)
```

```python
# Print the original array
print(arrB)

# Create a boolean mask for values less than 7
z = arrB < 7

# Filter arrB using the mask
arrC = arrB[z]

print(arrC)
```

```python
# Create a boolean mask for values >= 7
y = arrB >= 7

# Filter arrB using the new mask
arrD = arrB[y]

print(arrD)
```
