## Mathematical Operations

```python
# Print r_1 and its shape to understand its size and structure
print("r_1:", r_1)
print("Shape of r_1:", r_1.shape)  # .shape tells us how many rows and columns

# Print r_2 and its shape
print("r_2:", r_2)
print("Shape of r_2:", r_2.shape)

# Use np.add() to add the arrays element-by-element
# np.add is used instead of + because it makes broadcasting behavior explicit
result = np.add(r_1, r_2)

# Print the result and its shape
print("Result of r_1 + r_2:", result)
print("Shape of result:", result.shape)
```

```python
# Print starting values and shapes
print ("r_1:", r_1)
print ("Shape of r_1:", r_1.shape)

print("r_3:", r_3)
print("Shape of r_3:", r_3.shape)

# Add r_1 and r_3 using np.add
# NumPy attempts broadcasting to make shapes compatible
result = np.add(r_1, r_3)

print("Result of r_1 + r_2:", result)
print("Shape of result:", result.shape)
```

```python
# Print arrays and shapes before operation
print("r_2:", r_2)
print("Shape of r_2:", r_2.shape)

print("r_4:", r_4)
print("Shape of r_4:", r_4.shape)

# Add arrays using broadcasting rules
# r_2 will be broadcast across rows of r_4
result = np.add(r_2, r_4)

print("Result of r_2 + r_4:", result)
print("Shape of result:", result.shape)

# r_2 has shape (1,2) and r_4 has shape (4,5)
# Broadcasting fails because the number of columns (2 vs 5) does not match
# NumPy requires dimensions to be equal or one of them to be 1
```

```python
# Create an expression to illustrate what happens when you add r_1 and r_4
# This operation will raise an error because r_1 and r_4
# have incompatible shapes for broadcasting
np.add(r_1, r_4)

# NumPy cannot stretch these arrays to match each other
```

```python
# Show what happens if you add np.array([1]) to r_4.
print(np.add(np.array([1]),r_4,))

# np.array([1]) acts like a single number
# It is added to every value in r_4
```

```python
# Show what happens if you add np.array([[1]]) to r_4.
#print(np.add(np.array([[1]]),r_4,))
print(np.add(r_4, np.array([[1]])))


# The shape stays the same
# The array layout stays the same
# Only the values increase by 1
# NumPy expands [[1]] to match the shape of r_4.
# Then 1 is added to every element in r_4
```

```python
# Show what happens if you add np.array([1,1,1,1,1]) to r_4.
print(np.add(np.array([1, 1, 1, 1, 1]), r_4))

# np.array([1, 1, 1, 1, 1]) → shape (1,5)
# r_4 → shape (4,5)
# NumPy is pretending the small array is bigger just for the calculation.
# This array has the same number of columns as r_4
# It gets added row-by-row across r_4
# The array is extended only for the calculation
# r_4 keeps its original shape
```

```python
# What happens if you add np.array([[1],[1]]) to r_4? Add a comment to the
# cell with the error message this produces.
print(np.add(np.array([[1],[1]]), r_4))

# ValueError: operands could not be broadcast together with shapes (2,1) (4,5) 
# This is a column vector with shape (2,1)
# It does NOT match r_4's number of rows (4),
# so broadcasting fails
```

Broadcasting allows NumPy to perform operations on arrays with different shapes.

How it works:
- NumPy compares shapes from right to left
- Dimensions must match OR one of them must be 1
- If compatible, the smaller array is stretched to match the larger one

When using np.add:
- Values are added element-by-element
- The final shape is determined by the largest compatible shape

## Rounding

```python
# Print r_3 to display the values you’ll be working with for the next few cells.
print(r_3)
```

```python
# Round the values in r_3 using trunc.
print("trunc:\n", np.trunc(r_3))

# trunc removes the decimal part and keeps only the integer part
# It simply cuts off the number (no rounding up or down)
```

```python
# Round the values in r_3 using fix.
print("fix:\n", np.fix(r_3))

# fix works like trunc for positive numbers
# It removes the decimal part (towards zero)
```

```python
# Round the values in r_3 using floor.
print("floor:\n", np.floor(r_3))

# floor always rounds DOWN to the nearest whole number
# Since r_3 values are between 0 and 1, result will always be 0
```

```python
# Round the values in r_3 using ceil.
print("ceil:\n", np.ceil(r_3))
    
# ceil always rounds UP to the nearest whole number
# Since r_3 values are between 0 and 1, result will always be 1
```

```python
# Round the values in r_3 using round, to 2 decimal places.
print("round (2 decimals):\n", np.round(r_3, 2))

# round() rounds numbers to a specific number of decimal places
# Here we keep only 2 decimal places
```

## Cumulative Operations 

```python
# Print r_4 to display the values you’ll be working with for the next few cells.
print(r_4)
```

```python
# Print the sum of all values in r_4.
total = np.sum(r_4)
print("Total sum of all elements in r_4:", total)
```

```python
# Print the sum of elements along the 0 axis for r_4.
# axis=0 means column-wise sum (top to bottom)
col_sum = np.sum(r_4, axis=0)

print("Sum along axis 0 (columns):", col_sum)
```

```python
# Print the sum of elements along the 1 axis for r_4.
# axis=1 means row-wise sum (left to right)
row_sum = np.sum(r_4, axis=1)

print("Sum along axis 1 (rows):", row_sum)
```

```python
# Print the cumulative sum (rolling total) of values in r_4.
# cumsum adds values step-by-step in a single line (flattened array)
cum_total = np.cumsum(r_4)
print("Cumulative sum (flattened):", cum_total)
```

```python
# Print the cumulative sum (rolling total) of values along the 0 axis for r_4.
# cumulative sum column by column
# cumsum with axis=0 adds values down each column (top to bottom)
cum_axis0 = np.cumsum(r_4, axis=0)
print("Cumulative sum along axis 0:\n", cum_axis0)
```

## Grocery Problem

```python
# Create two new variables, bag1 and bag2
# Order: [vegetables, fruits, snacks, beverages]

bag1 = np.array([2, 3, 4, 0])  # 2 veg, 3 fruit, 4 snacks, 0 drinks
bag2 = np.array([4, 0, 1, 2])  # 4 veg, 0 fruit, 1 snack, 2 drinks

print("Bag 1:", bag1)
print("Bag 2:", bag2)
```

```python
# Sum each bag separately
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
# Add bag1 and bag2 element-wise
totals = bag1 + bag2

print(
    f"There are {totals[0]} vegetables, {totals[1]} fruits, {totals[2]} boxes of snacks, and {totals[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
# Ali has full 5 days of data (Mon–Fri)
ali = np.array([5, 4, 6, 4, 2])

# Ben missed Monday, so we use 0 or we just start from Tuesday
ben = np.array([0, 5, 5, 3, 4])

# Carla missed Wednesday, so we put 0 for that day
carla = np.array([3, 6, 0, 3, 4])

print("Ali:", ali)
print("Ben:", ben)
print("Carla:", carla)
```

```python
# Stack all people into one 2D array (rows = people, columns = days)
all_knocks = np.array([ali, ben, carla])

print("All knocks:\n", all_knocks)
```

```python
# Sum across days (axis=1 → row-wise)
# This gives total for each person
person_totals = np.sum(all_knocks, axis=1)

# Total for the whole group
group_total = np.sum(all_knocks)

print("Total per person:", person_totals)
print("Group total:", group_total)
```

```python
# Sum across people (axis=0 → column-wise per day)
daily_total = np.sum(all_knocks, axis=0)

print("Daily total knocks:", daily_total)

# Cumulative (rolling) total across the week
rolling_knocks = np.cumsum(daily_total)

print("Rolling total knocks:", rolling_knocks)
```

```python
# Every 2 knocks = 1 vote → multiply by 0.5
estimated_votes = rolling_knocks * 0.5

print("Estimated rolling votes:", estimated_votes)
```
