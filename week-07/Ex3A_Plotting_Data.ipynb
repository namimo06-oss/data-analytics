# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
# Create two variables 
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
# Use plt.plot() to plot the line graph for x and y, and then display it using plt.show()
plt.plot(x,y)
plt.show()
```

```python
# Duplicate the previous code cell, then update it to add markers to the data points using the marker='o' parameter
# Plotting the line graph with markers on each data point
plt.plot(x, y, marker='o')

# Displaying the plot
plt.show()
```

```python
# Duplicate the previous code cell, then customize the plot with a title (‘Example
# Graph’), x-axis label (‘X axis’), and y-axis label (‘Y axis’)
# Plotting the line graph with markers
plt.plot(x, y, marker='o')

# Adding a title and axis labels
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

# Displaying the plot
plt.show()
```

## Styling Plots 

```python
# Plot the same data as above, but this time, change the line style to dashed ('--') and
# the line color to red.
plt.plot(x, y, linestyle='--', color='red')

# Displaying the plot
plt.show()
```

```python
# Change the marker style to a square ('s'), set the marker size to 10, and change the
# marker color to green.
# Plotting with square markers, larger size, and green color
plt.plot(x, y, marker='s', markersize=10, markerfacecolor='green')

# Displaying the plot
plt.show()
```

```python
# Experiment with at least two different additional marker styles and colors, and
# display the plots
# First additional marker style
plt.plot(x, y, marker='^', color='blue')
plt.show()

# Second additional marker style
plt.plot(x, y, marker='D', color='purple')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques 

```python
# Data for the bar chart
# 'products' represents the categories on the x-axis
# 'sales' represents the numerical values (heights of the bars)
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

# Create a bar chart
# plt.bar() is used specifically for bar charts
plt.bar(products, sales)

# Display the plot
plt.show()
```

```python
# Customize your bar chart by adding a title, x-axis label, and y-axis label. Set the
# color of the bars to 'skyblue' and adjust the bar width to 0.5
plt.bar(products, sales, color='skyblue', width= 0.5)

plt.title('Sales by Product')
plt.xlabel('Products')
plt.ylabel('Sale')

# Display the plot
plt.show()
```

```python
# Create a scatter plot to visualize the relationship between two variables. Use the
# following data:
x = np.random.rand(50)
y = np.random.rand(50)

# Create a scatter plot
# plt.scatter() is used to show relationships between two variables
plt.scatter(x, y)
```

```python
# Customize your scatter plot by setting the size of the markers based on another
# variable (e.g., s = np.random.rand(50) * 100), and using a colormap (e.g.,
# cmap='viridis')
s = np.random.rand(50) * 100
plt.scatter(x, y, s=s, c=s, cmap='viridis')
plt.colorbar()
```

## Subplots

```python
# Create a figure with 2 subplots (1 row, 2 columns)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
```

```python
# Subplots With Line Chart And Histogram
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# First Subplot
axes[0].plot(x, y)
axes[0].set_title('Sine Wave')
axes[0].set_xlabel('X values')
axes[0].set_ylabel('Y values')

# Second Subplot
random_numbers = np.random.randn(1000)

axes[1].hist(random_numbers)
axes[1].set_title('Histogram')
axes[1].set_xlabel('Values')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
