# Exercise 2.B Cleaning up data 

```python
# Import numpy and pandas 
import numpy as np
import pandas as pd 
```

```python
# In a code cell, create a new DataFrame named df based on test_data.csv (as saved
# from the previous exercise)
df = pd.read_csv('test_data.csv')
df
```

```python
# Add a cell to show column statistics on df using the describe method
df.describe()
```

## Filtering a DataFrame 

```python
# Filter the DataFrame to show only rows where Age is greater than 18.
df_age_over_18 = df[df['Age'] > 18]

df_age_over_18
```

```python
# Filter the DataFrame to show only rows where Score is greater than 50 and less than 100
df_score_range = df[(df['Score'] > 50) & (df['Score'] < 100)]

df_score_range
```

## Modifying a DataFrame

```python
# Create new variable df_cleaned. Assign it the value of df, using the drop function to remove rows where the Score is less than 0 
#(i.e., invalid score data). Then call df_cleaned to review the result

# We create a new DataFrame called df_cleaned
# drop() is used to remove rows that meet a condition
# Here: remove rows where Score < 0 (invalid data)

df_cleaned = df.drop(df[df["Score"] < 0].index)

# Display the cleaned DataFrame
df_cleaned
```

```python
# Continuing to work with df_cleaned for the rest of these steps, remove the City
# column from the DataFrame using drop. Call df_cleaned to review the result

# We continue working with df_cleaned
# axis=1 means we are removing a column (not rows)

df_cleaned = df_cleaned.drop("City", axis=1)

# Display the DataFrame after removing City
df_cleaned
```

```python
# Replace all missing Age values with the mean Age of the DataFrame using fillna.
# Call df_cleaned to review the result

# Calculate the mean (average) age, ignoring NaN automatically
mean_age = df_cleaned["Age"].mean()

# fillna() replaces NaN values with the mean age
df_cleaned["Age"] = df_cleaned["Age"].fillna(mean_age)

# Display the updated DataFrame
df_cleaned
```

```python
# Replace any Score value of 0 with the mean Score of the DataFrame using replace.
# Call df_cleaned to review the result.
mean_score = df_cleaned["Score"].mean() # Calculate the mean Score

df_cleaned["Score"] = df_cleaned["Score"].replace(0, mean_score)  # replace() changes Score values of 0 into the mean Score

df_cleaned
```

```python
# Remove any remaining rows that contain NaN values using the dropna function.
# Call df_cleaned to review the result.

df_cleaned = df_cleaned.dropna() # dropna() removes any row that still has NaN values

# Display the cleaned DataFrame
df_cleaned
```

```python
# Apply the values from ID as index labels, then remove column ID. (Pay attention to
# the numbering!) Call df_cleaned to review the result

# set_index() moves the ID column to be the index (row labels)
df_cleaned = df_cleaned.set_index("ID")

# Display the final cleaned DataFrame
df_cleaned
```
