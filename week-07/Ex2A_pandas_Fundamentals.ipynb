## Series


```python
# Create four new variables 
aa = [1, 2, 3, 4]  # list []
ab = (5, 6, 7, 8)  # tuple ()
ac = {9, 10, 11, 12} # set{} # NOTE: sets are unordered (important for pandas behavior)
ad = {   # dictionarykey/value pairs 
    "first": 13,
    "second": 14,
    "third": 15,
    "fourth": 16
}

print("aa:", aa)
print("ab:", ab)
print("ac:", ac)
print("ad:", ad)
```

```python
# Convert Python objects into pandas Series

ser1 = pd.Series(aa)  # from list
ser2 = pd.Series(ab)  # from tuple

# needs to be cast as a list otherwise will encounter a typeerror saying that the set type is unordered
ser3 = pd.Series(list(ac)) # from set (order may change!)
ser4 = pd.Series(ad)  # from dictionary (keys become index)

# in "" I add a small subtitle to know what is it showing
print("ser1:\n", ser1)
print("\nser2:\n", ser2)
print("\nser3:\n", ser3)
print("\nser4:\n", ser4)

```

```python
# This will cause an error because pandas expects a list-like object
ser5 = pd.Series(17, 18, 19, 20)
```

## Dataframes from Series 

```python
# Create a DataFrame using ser1
# When a Series is used to create a DataFrame,
# it becomes a single column DataFrame
df1 = pd.DataFrame(ser1)

# Calling the variable name (not using print)
# Jupyter will display it as a formatted table
df1
```

```python
# insert() adds a new column at a specific position
# Syntax: df.insert(column_position, column_name, column_values)

# Insert ser2 as a new column at position 1
df1.insert(1, "col_from_ser2", ser2)

# Insert ser3 as a new column at position 2
df1.insert(2, "col_from_ser3", ser3)

# View the updated DataFrame
df1
```

```python
# Try to add ser4 as a new column
df1["col_from_ser4"] = ser4

# View the DataFrame
df1
```

```python
# Add a new row with index label 4
# .loc allows you to assign values using a row label
df1.loc[4] = [5, 9, 12, 0]

# View the updated DataFrame
df1
```

### Facts from df2.info()

From the info output, we can determine:

```python
# Show summary statistics for numeric columns
df2.describe()
```

### Insights from df2.describe()

Useful insights:
- Typical (average) values in each numeric column
- Spread of the data (standard deviation)
- Minimum and maximum values

Less useful summarizations:
- For ID columns, the mean and quartiles are often not meaningful
- Statistics on coded identifiers don’t describe real-world behavior

## Coding a DataFrame

```python
# Create a dictionary
# Each key will become a column name in the DataFrame
# Each value (list) must be the same length
test_data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Name': ['Alicia', 'Bob', 'Charlie', 'Dominic', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],
    'City': ['NY', 'LA', 'Chicago', 'Houston', 'Phoenix', 'Boston', 'Austin', 'San Diego'],
    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
# Create a DataFrame from the dictionary
df4 = pd.DataFrame(test_data)

# Display the DataFrame
df4
```

```python
# Export df4 to a CSV file
# index=False prevents pandas from saving the index as a column
df4.to_csv('test_data.csv', index=False)
```
