# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
# Create a DataFrame named df_sales using the following data:
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}

# Convert dictionary into DataFrame
df_sales = pd.DataFrame(data)

# Display DataFrame
df_sales
```

```python
# Add a new column Revenue to df_sales by multiplying Units_Sold and Price.

# Revenue = Units_Sold × Price
df_sales["Revenue"] = df_sales["Units_Sold"] * df_sales["Price"]

# Show updated DataFrame
df_sales
```

```python
# Use the sum function to calculate the total Units_Sold and total Revenue across all products and regions.

# Sum of all Units_Sold
total_units = df_sales["Units_Sold"].sum()

# Sum of all Revenue
total_revenue = df_sales["Revenue"].sum()

# Print results
print("Total Units Sold:", total_units)
print("Total Revenue:", total_revenue)
```

```python
# Use the mean function to calculate the average Units_Sold and average Revenue

# Average Units Sold
avg_units = df_sales["Units_Sold"].mean().round(2)

# Average Revenue
avg_revenue = df_sales["Revenue"].mean().round(2)

# Print results
print("Average Units Sold:", avg_units)
print("Average Revenue:", avg_revenue)
```

## Grouping Data

```python
# Use the groupby function to group df_sales by Product and calculate the total Units_Sold and total Revenue for each product.
# Group by Product and calculate total Units_Sold and Revenue
# This means: combine all rows that have the same Product

df_sales.groupby("Product")[["Units_Sold", "Revenue"]].sum()
```

```python
# Use the groupby function to group df_sales by Region and calculate the average Units_Sold and average Revenue for each region

# Group by Region and calculate average values
# This shows typical performance per region

df_sales.groupby("Region")[["Units_Sold", "Revenue"]].mean()

```

```python
# Use the groupby function to group df_sales by both Product and Region and calculate the sum of Units_Sold and Revenue for each combination.
# Group by BOTH Product and Region
# This creates subgroups (like a matrix)

df_sales.groupby(["Product", "Region"])[["Units_Sold", "Revenue"]].sum()
```

## Sorting and Ranking Data

```python
# Sort from smallest to largest Units_Sold
# ascending=True is default, but we write it for clarity
df_sales.sort_values(by="Units_Sold", ascending=True)
```

```python
#Sort df_sales by Revenue in descending order.
# Sort from highest to lowest Revenue
df_sales.sort_values(by="Revenue", ascending=False)
```

```python
 # Add a new column Revenue_Rank to df_sales, which ranks the Revenue values in descending order (i.e., highest revenue has rank 1
# Rank Revenue in descending order
# highest revenue = rank 1
df_sales['Rank'] = df_sales['Revenue'].rank(ascending=False,method='dense').astype(int)
# Show updated DataFrame
df_sales
```

```python
# Show final DataFrame with all changes
df_sales
```

## Merging DataFrames 

```python
# Create a new DataFrame named df_products with the following data:
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
}

df_products = pd.DataFrame(product_data)

# Show the DataFrame
df_products
```

```python
# Merge df_sales with df_products on the Product column, resulting in a new DataFrame named df_merged
# Merge df_sales and df_products using the "Product" column
# This combines information from both tables

df_merged = pd.merge(df_sales, df_products, on="Product")

# Show result
df_merged
```

```python
# Just display the merged result

df_merged
```

```python
# Calculate total Revenue per Category

df_merged.groupby("Category")["Revenue"].sum()
```
