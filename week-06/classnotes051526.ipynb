## Error Handling
4.B.1 Handling errors / Working with try / except

```python

# without try / except
first_num = float(input('% to calculate as decimal (provide a whole number): '))

pct_as_dec = first_num / 100

print(pct_as_dec)
```

```python
# with try / except

try:
    first_num = float(input('% to calculate as decimal (provide a whole number): '))
    pct_as_dec = first_num / 100
except:
    print('Invalid input. Please try again')
else:
    print(pct_as_dec)
finally:
    print('Run again to calculate another decimal value, ')
```

```python
try:
    fnum = float(input('Number to divide: '))
    snum = float(input('Second number to use as divisor: '))
    div_result = fnum / snum
except ValueError:
    print('Invalid input: must be a number. Please try again')
except ZeroDivisionError:
    print('Cannot divide by zero.')
except:
    print('Encountered an error.' )
else:
    print(div_result)
finally:
    print('Run again to calculate another decimal value. ')
```

```python
# another alternative 
try:
    fnum = float(input('Number to divide: '))
    snum = float(input('Second number to use as divisor: '))
    div_result = fnum / snum
except (ValueError, ZeroDivisionError): # this is also a valid soluction to the same problem
    print('Error: must be a number other than zero')
except:
    print('Encountered an error.' )
else:
    print(div_result)
finally:
    print('Run again to calculate another decimal value. ')
```

### using try / except with a lop PG 75

```python
print('Division calculator:')
first_n = None

while first_n == None:
    try:
        first_n = float(input('First number to divide: '))
    except:
        print('Invalid input! Must be a number')

second_n = None
answer = None

while second_n == None:
    try:
        second_n = float(input('Second number to divide by: '))
        answer = first_n / second_n
    except ValueError:
        print('Must be a number')
    except ZeroDivisionError:
        print('Number cannot be zero')
        second_n = None  # so it can go back and ask for second number again 

print(answer)
        
```

```python
print('Division calculator:')
first_n = None

while first_n == None:
    try:
        first_n = float(input('First number to divide: '))
    except:
        print('Invalid input! Must be a number')

second_n = None
answer = None

while second_n == None:
    try:
        second_n = float(input('Second number to divide by: ')) #this woudl not result in an error if the number is zero  
        answer = first_n / second_n
    except: 
        pass # would terminate the lop so you get end of program, but it's less informative
print(answer)
        
```

```python
#another example of try / except combined with a loop that's operatingon file contents

try:
    with open("sample.txt.txt") as f:  # if no sample2.txt then will trigger except block
        contents = f.readlines() 
except FileNotFoundError:
    print('Error:file not found')
else:
    with open("sales_data.csv", "a") as output:
        for i in contents:
            if "," in i:
                output.write(i)
    print('Output complete.')
finally:
    print('Run program again to process another file.')
```
